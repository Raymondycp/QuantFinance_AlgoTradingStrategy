# EMA Cross Strategy (M1)
This is a systematic trend-following(EMA) approach with controlled entries (waiting for pullbacks) and multiple exit options for flexibility in different market conditions.

## 1. Follow the Trend (EMA)
Trend Identification: Uses multiple EMAs (7, 20, 50 periods) to determine market direction

EMA Hierarchy:
Bullish Trend: When EMA7 > EMA20 > EMA50 (with price closed > EMA7 for confirmation)
Bearish Trend: When EMA7 < EMA20 < EMA50 (with price closed < EMA7 for confirmation)
Slope Confirmation: Checks if EMAs are rising (for uptrend) or falling (for downtrend)

## 2. Pullback for Entry
Entry Signals:
Long Entry: Price pull back EMA20 during Bullish Trend
Short Entry: Price pull back EMA20 during Bearish Trend
Filter: Price close < EMA800

## 3. Exit Options (Flexible)
Primary Exit Methods:
Fixed percentage stop loss (1% default)
Fixed percentage take profit (1% default)

Passive Optional 1: ATR-based dynamic take profit
Passive Optional 2: EMA Cross Exit Rule:
Closes longs when price closed < EMA7
Closes shorts when price closed > EMA7

## Additional Features
Bollinger Bands: Visual reference for volatility (not used in logic)
RSI filter can be disabled
